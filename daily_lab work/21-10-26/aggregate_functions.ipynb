```python
# Search , Sort and Aggregate / Statistics Functions
import numpy as np

numbers = np.array([25 , 42 , 13 , 3 , 31])

print("Numbers: ", numbers)
print("Max Value: ", np.max(numbers))
print("Min Value: ", np.min(numbers))
print("Index of Max Value: ", np.argmax (numbers))
print("Index of Min Value: ", np.argmin(numbers))
```

```python
# Sort 2D array along diffrence axes

matrix = np.array([
    [9 , 2 , 3],
    [4 , 8 , 6],
    [1 , 5 , 7]
])

print("Matrix: \n", matrix)
print("Sorted Matrix along axis 0: \n", np.sort(matrix, axis=0))
print("Sorted Matrix along axis 1: \n", np.sort(matrix, axis=1))
```

```python
# Find Mean  , median , varience and standard deviation

scorces = np.array([25 , 42 , 13 , 3 , 31])
print("Scorces: ", scorces)
print("Mean: ", np.mean(scorces))
print("Median: ", np.median(scorces))
print("Variance: ", np.var(scorces))
print("Standard Deviation: ", np.std(scorces))
```

```python
# Numpy: Filtering , Masking , Reshaping & Broadcasting

ages = np.array([28 , 32 , 35 , 26 , 17 , 12])

adults = ages[ages >= 18]
print("Ages", ages)
print("Adults", adults)
```

```python
young_ages = ages[ages < 18]
print("Young Ages", young_ages)
```

```python
# Filtering String

strings = np.array(['apple', 'banana', 'cherry', 'date', 'elderberry'])

selected_strings = strings[np.char.startswith(strings, 'b')]
print("Strings: ", strings)
print("Selected Strings: ", selected_strings)
```
