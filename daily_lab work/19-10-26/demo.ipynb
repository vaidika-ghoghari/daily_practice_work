```python
# print 0 to 9 numbers using numpy
import numpy as np
arr =  np.arange(0,10 , 1)

print(arr)
print(arr.shape)
```

```python
# Generate a 2D Array with random values and print shape and datatypes

arr = np.random.rand(3 , 4)
print(arr)
print(arr.shape)
print(arr.dtype)
```

```python
# Create 50 Evenly spaced values between 0 to 10

arr = np.linspace(0 , 10 , 50)

print(arr)
```

```python
# np.zeros

arr = np.zeros((3,4))

print(arr)
```

```python
# np.ones

arr = np.ones((3,4))
print(arr)
```

```python
# Generate a 3x3 identity matrix using NumPy.

arr = np.eye(3)

print(arr)
print(arr.shape)
```

```python
# Create a 4x4 array filled with random integers between 1 and 20.

arr = np.random.randint(1 , 21 , size =(4,4))
print(arr)
```

```python
# Create a 1D NumPy array with values ranging from 1 to 15. Slice the array to extract elements from index 3 to 8.

arr = np.arange(1 , 16)

print(arr)

print("Sliced array from index 3 to 8:", arr[3:9])
print("Sliced array from reverse order:", arr[::-1])
print("Sliced array from index 3 to end:", arr[-3:])
print("Sliced array from start to index 5:", arr[:5])
```

```python
# Generate a 3x3 matrix with values ranging from 1 to 9. Retrieve the element at the second row and third column.

arr = np.arange(1 , 10).reshape(3,3)

print(arr)

print("Element at second row and third column:", arr[1,2])
```

```python
# Slice a 4x4 array to extract the last two rows and the last two columns.

arr = np.arange(1 , 17).reshape(4,4)

print(arr)

print("Sliced Matrix for last two rows and last two columns:\n", arr[-2:,-2:])
```
