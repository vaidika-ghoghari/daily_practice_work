```python
import pandas as pd 
import numpy as np
```

```python
# Dictionary of lists


students = pd.DataFrame({
    'name': ['Alice', 'Bob', 'Charlie', 'Diana', 'Ethan'],
    'city': ['New York', 'London', 'Paris', 'Tokyo', 'Sydney'],
    'course': ['Computer Science', 'Mathematics', 'Physics', 'Biology', 'Biology'],
    'marks': [85, 92, 78, 88, 95],
    'fees': [5000.00, 4500.00, 5000.00, 6000.00, 5500.00]
})

print(students)
```

```python
print("Head of 2:\n",students["name"].head(2))

print("\nHead with arguments:\n",students[["name" , "marks"]].head(2))

print("\nloc:\n",students.loc[2])

print("\nloc with arguments:\n",students.loc[0:2 ,["name" ,"marks"]])

```

```python
# Add a new row  , reset the row numbers

new_student = pd.DataFrame([{"name":"Mohit" , "city":"surat" , "course":"python" , "marks":97 , "fees":20000}])

print(new_student)
```

```python
students = pd.concat([students,new_student] , ignore_index=True)
print(students)
```

```python
students["fees_paid"] = 20000 - students["fees"]

students["result"] = np.where(students["marks"] >= 60 ,"pass" , "fail")

print(students)
```

```python
# Rename columns

print(students.rename(columns={"name":"Name"}))
```

```python
#remove col

print(students.drop(columns="result"))

```
