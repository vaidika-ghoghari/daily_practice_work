```python
#Task 2.1: Basic Masking for Selection
#Create a NumPy array temperatures with the following values: [28, 32, 25, 35, 30, 22, 38].
#reate a boolean mask is_hot where True indicates temperatures above 30 degrees Celsius.
#Use this mask to select the hot temperatures from the temperatures array.
```

```python
import numpy as np

temperatures = np.array([28 , 32 , 25 , 35 , 30 , 22 , 38])

is_hot = temperatures > 30  

hot_temperatures = temperatures[is_hot]
print(temperatures)
print(is_hot)
print("Hot Tempretures :" ,hot_temperatures)
```

```python
#Task 2.2: Masking for Modification Using the temperatures array again, use the is_hot mask to change all temperatures above 30 degrees Celsius to 31 degrees Celsius.
temperatures[is_hot] = 31

print(temperatures)
```

```python
#Using np.where() for Masking Create a NumPy array scores with the following values: [65, 80, 72, 90, 55, 78].Use np.where() to create a new array grades. If a score is 75 or above, the grade should be 'Pass', otherwise 'Fail'.

scores = np.array([65 , 80 , 72 , 90 , 55 , 78])
grades = np.where((scores > 75) , "Pass" , "Fail")
result = scores[grades == "Pass"]
print(grades)
print("Passing grades",result)
```

```python
#Reshaping a 1D array to 2D Create a 1D NumPy array numbers_1d with values from 1 to 12 (inclusive).Reshape this array into a 2D array with 3 rows and 4 columns.
import numpy as np

numbers = np.array([2 , 5 , 6 , 8 , 3 , 9 ,1 , 7 , 4 ,11 , 10 ,12])

numbers_1d = numbers.reshape(12)
numbers_2d = numbers.reshape(3 , 4)
print("number_1d",numbers_1d)
print("number_2d\n",numbers_2d)
```

```python
numbers_1d = numbers.flatten()
print(numbers_1d)

numbers_2d = numbers.reshape(6 , -1)
print(numbers_2d)
```
