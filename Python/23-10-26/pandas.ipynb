```python
import pandas as pd
import numpy as np
```

```python
# A Series from a list. Pandas gives defults labels 0 , 1 , 2......

marks = pd.Series([78 ,92 ,85 , 65]  , index = ["Jhon", "peter" ,"Alice" ,"bob"])
print("marks:\n", marks)
```

```python
print("Alice's Marks: ", marks["Alice"])#by ladel
print("Jhon's Marks: ", marks["Jhon"])#by ladel
```

```python
# A series from a dictionary:keys become labels
fees = pd.Series({"HTML":30000 , "Python":50000 , "Javascript":100000})

print(fees)
```

```python
#Operations
print(marks + 5 )
print()
print(marks[marks >= 80])
print()
print("Max :",marks.max() )
print("Min :",marks.min() )
print("Mean:",marks.mean())
```

```python
# DataFrames: 4 Ways

# Dictionary of lists

students = pd.DataFrame({
    'name': ['Alice', 'Bob', 'Charlie', 'Diana', 'Ethan'],
    'city': ['New York', 'London', 'Paris', 'Tokyo', 'Sydney'],
    'course': ['Computer Science', 'Mathematics', 'Physics', 'Biology', 'Biology'],
    'marks': [85, 92, 78, 88, 95],
    'fees': [5000.00, 4500.00, 5000.00, 6000.00, 5500.00]
})

print(students)
```

```python
#list of list + columns
data = [
    [101, "Amit", 25000],
    [102, "Neha", 30000],
    [103, "Rahul", 28000]
]

columns = ["ID","NAME","SALARY"]

emp = pd.DataFrame(data , columns=columns)
print(emp)
```

```python
#list of dictionaries
data = [
    {"ID": 101, "Name": "Amit", "Salary": 25000},
    {"ID": 102, "Name": "Neha", "Salary": 30000},
    {"ID": 103, "Name": "Rahul", "Salary": 28000}
]

emp = pd.DataFrame(data)
print(emp)
```

```python
#combine series into dataframe

combo = pd.DataFrame({"Marks" :marks , "Bonus":[2 , 0 , 5 ,1]})
print(combo)
```
