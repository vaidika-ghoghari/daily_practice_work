```python
# Broadcasting a Scalar to an array
import numpy as np  
```

```python
prices = np.array([10 , 20 , 30 , 40 , 50])
print("Original prices ",prices)

new_prices = prices + 5
print("New prices",new_prices)
```

```python
# Brodcasting 1D array to a 2D array

sales = np.array([10 ,20 ,30 ,10 ,30 , 20])
sales_2d = sales.reshape(3,2)
print(sales_2d)
```

```python
bonuses = np.array([2 , 3 , 1])
bonuses_col = bonuses.reshape(3,1)
print(bonuses)
print(bonuses_col)
```

```python
sales_with_bonus = sales_2d + bonuses_col

print(sales_with_bonus)
```
